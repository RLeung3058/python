```python
# Filter the DataFrame to show only rows where station is "COLLEGE STATION " or "ROYAL YORK STATION"

stations_of_interest = ['COLLEGE STATION', 'ROYAL YORK STATION']

# filter using .loc[] and .isin()
filtered_stations = df.loc[df["station"].isin(stations_of_interest)]

print(filtered_stations)
```
